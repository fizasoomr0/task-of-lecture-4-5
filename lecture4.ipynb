```python
# Task 1:
# Print all numbers divisible by 7 from 1 to 100.

for number in range(1, 101):
    if number % 7 == 0:
        print(number)
```

```python
# Task 2:
# Calculate the sum of even numbers from 1 to 50.

total = 0

for number in range(1, 51):
    if number % 2 == 0:
        total += number

print("Sum of even numbers:", total)
```

```python
# Task 4:
# Create a countdown from 10 to 1.
# Stop when the countdown reaches 5.
# Use break and print "Midpoint reached".

number = 10

while number >= 1:
    print(number)

    if number == 5:
        print("Midpoint reached")
        break

    number -= 1
```

```python
# Task 6:
# Create a dictionary where:
# key = word
# value = length of word
# Only include words longer than 4 letters.

words = ["apple", "cat", "banana", "book", "python", "fiza", "computer"]

word_dictionary = {
    word: len(word)
    for word in words
    if len(word) > 4
}

print(word_dictionary)
```

```python
# Task 7:
# Given a string, create a dictionary
# where keys are characters
# and values are their frequency.

text = "programming"

frequency = {
    character: text.count(character)
    for character in set(text)
}

print(frequency)
```

```python
# Task 8:
# Create a list from 100 to 150.
# Prime numbers remain as numbers.
# Non-prime numbers become "Not Prime".

def is_prime(number):

    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True


result = [
    number if is_prime(number) else "Not Prime"
    for number in range(100, 151)
]

print(result)
```

```python
# ==========================================
# Student Grade System
# ==========================================

students = []

while True:

    # Get student name
    name = input("\nEnter student name (or 'done' to finish): ").strip()

    # Stop entering students
    if name.lower() == "done":
        break

    print(f"\nEnter marks for {name}:")

    # Get marks for five subjects
    english = float(input("English: "))
    mathematics = float(input("Mathematics: "))
    programming = float(input("Programming: "))
    database = float(input("Database: "))
    communication = float(input("Communication: "))

    # Store all marks in a list
    marks = [
        english,
        mathematics,
        programming,
        database,
        communication
    ]

    # Check for invalid marks
    if any(mark < 0 or mark > 100 for mark in marks):
        print("Invalid marks! Each mark must be between 0 and 100.")
        continue

    # Calculate total and average
    total_marks = sum(marks)
    average_marks = total_marks / len(marks)

    # Store student record
    student = {
        "name": name,
        "marks": marks,
        "total": total_marks,
        "average": average_marks
    }

    students.append(student)

    print("Student record added successfully.")


# ==========================================
# Display Student Results
# ==========================================

if students:

    print("\n==============================================")
    print("              STUDENT GRADE SYSTEM")
    print("==============================================")

    for student in students:

        print(f"\nStudent Name: {student['name']}")
        print("----------------------------------------------")

        print(f"English          : {student['marks'][0]}")
        print(f"Mathematics      : {student['marks'][1]}")
        print(f"Programming      : {student['marks'][2]}")
        print(f"Database         : {student['marks'][3]}")
        print(f"Communication    : {student['marks'][4]}")

        print("----------------------------------------------")
        print(f"Total Marks      : {student['total']}")
        print(f"Average Marks    : {student['average']:.2f}")


    # ==========================================
    # Overall Class Results
    # ==========================================

    averages = [student["average"] for student in students]

    highest_average = max(averages)
    lowest_average = min(averages)
    overall_average = sum(averages) / len(averages)
    total_students = len(students)

    print("\n==============================================")
    print("                OVERALL RESULTS")
    print("==============================================")

    print(f"Total Students   : {total_students}")
    print(f"Overall Average  : {overall_average:.2f}")
    print(f"Highest Average  : {highest_average:.2f}")
    print(f"Lowest Average   : {lowest_average:.2f}")

    print("==============================================")

else:

    print("\nNo valid student records were entered.")
```

```python
# List Categorizer
# -----------------------------------
# Create a list from 1 to 20.
# Separate numbers into:
# 1. Even
# 2. Odd
# 3. Multiples of 5

l = list(range(1, 21))

even = []
odd = []
multiples_of_5 = []

for number in l:

    # Check for even numbers
    if number % 2 == 0:
        even.append(number)

    # Check for odd numbers
    if number % 2 != 0:
        odd.append(number)

    # Check for multiples of 5
    if number % 5 == 0:
        multiples_of_5.append(number)


# Print all three lists
print("Even:", even)
print("Odd:", odd)
print("Multiples of 5:", multiples_of_5)
```

```python
# FizzBuzz Dictionary
# -----------------------------------
# Create a dictionary for numbers 1 to 30.
#
# Divisible by both 3 and 5 = FizzBuzz
# Divisible by 3 = Fizz
# Divisible by 5 = Buzz
# Otherwise = None

fizzbuzz = {}

for number in range(1, 31):

    # Divisible by both 3 and 5
    if number % 3 == 0 and number % 5 == 0:
        fizzbuzz[number] = "FizzBuzz"

    # Divisible by 3
    elif number % 3 == 0:
        fizzbuzz[number] = "Fizz"

    # Divisible by 5
    elif number % 5 == 0:
        fizzbuzz[number] = "Buzz"

    # Not divisible by 3 or 5
    else:
        fizzbuzz[number] = "None"


print(fizzbuzz)
```
