```python
# Find the first number from 1 to 500
# that is divisible by both 13 and 7

for number in range(1, 501):
    if number % 13 == 0 and number % 7 == 0:
        print("First number divisible by both 13 and 7:", number)
        break
```

```python
# Count the frequency of each character in a string

text = "hello"

frequency = {
    character: text.count(character)
    for character in set(text)
}

print("Character Frequency:")
print(frequency)
```

```python
# Create a list from 100 to 150
# Prime numbers remain as they are
# Non-prime numbers become "Not Prime"

numbers = list(range(100, 151))


result = [
    number if number > 1 and all(number % i != 0 for i in range(2, int(number ** 0.5) + 1))
    else "Not Prime"
    for number in numbers
]


print("Prime Number List:")
print(result)
```

```python
# ==========================================
# Function to Find Prime Numbers Up to n
# ==========================================

def prime_upto(n):

    # Input type validation
    if not isinstance(n, int):
        return "Error: n must be an integer."

    # Input value validation
    if n < 2:
        return "Error: n must be greater than or equal to 2."

    primes = []

    # Check each number from 2 up to n
    for number in range(2, n + 1):

        is_prime = True

        # Check whether the number is divisible
        # by any number other than 1 and itself
        for divisor in range(2, int(number ** 0.5) + 1):
            if number % divisor == 0:
                is_prime = False
                break

        if is_prime:
            primes.append(number)

    return primes


# ==========================================
# Test the Function
# ==========================================

print("Prime numbers up to 50:")
print(prime_upto(50))
```

```python
# ==========================================
# Number Summary Function
# ==========================================

def number_summary(numbers):

    # Calculate total
    total = sum(numbers)

    # Calculate average
    average = total / len(numbers)

    # Find maximum and minimum values
    maximum = max(numbers)
    minimum = min(numbers)

    # Find all even numbers
    even_numbers = [number for number in numbers if number % 2 == 0]

    # Return results in a dictionary
    return {
        "total": total,
        "average": average,
        "max": maximum,
        "min": minimum,
        "even_numbers": even_numbers
    }


# ==========================================
# Test the Function
# ==========================================

numbers = [10, 15, 20, 25, 30, 35, 40]

result = number_summary(numbers)

print("Number Summary:")
print(result)
```

```python
# ==========================================
# Email Filtering System
# ==========================================

emails = [
    "fiza@company.com",
    "ayesha@gmail.com",
    "almira@company.com",
    "zara@company.com"
]

# Specific domain
domain = "@company.com"

# Extract emails from the specific domain
company_emails = [
    email for email in emails
    if email.endswith(domain)
]

# Extract unique usernames before '@'
unique_users = {
    email.split("@")[0]
    for email in company_emails
}

# ==========================================
# Display Results
# ==========================================

print("==========================================")
print("           EMAIL FILTERING SYSTEM")
print("==========================================")

print("\nEmails from", domain)
for email in company_emails:
    print("-", email)

print("\nUnique Users:")
for user in unique_users:
    print("-", user)

print("\nTotal Unique Users:", len(unique_users))

print("==========================================")
```
